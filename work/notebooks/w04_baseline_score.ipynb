## 1. My rule and its reason codes

**Baseline Scoring Rule:**

We prioritize content pieces that show high historical impression growth combined with declining recent engagement velocity, targeting them for content refreshment.

**Reason Codes:**

- HIGH_TRAFFIC_DECAY: Significant drop in recent clicks despite strong historic baseline.

- STAGNANT_ENGAGEMENT: Impressions remain high, but scroll depth and time-on-page metrics have plateaued.

- OPTIMIZATION_WINDOW: Content matches keywords with active search volume surges.

```python
# Code for Section 1: Rule definition validation
import pandas as pd

def validate_rule_parameters():
    reason_codes = ['HIGH_TRAFFIC_DECAY', 'STAGNANT_ENGAGEMENT', 'OPTIMIZATION_WINDOW']
    assert len(reason_codes) == 3, "Rule reason codes definition incomplete."
    print("Baseline rule and reason codes successfully validated.")

validate_rule_parameters()
```

## 2. Build the ranked queue (writes the CSV)

*Code the score, rank everything, write work/outputs/baseline_action_score.csv.*

**Ranked Queue Generation:**

We calculate a composite baseline action score using historical traffic decay and search volume metrics, sort the dataset in descending order, and export the prioritized output queue to work/outputs/baseline_action_score.csv.

```python
# Code for Section 2: Build score, rank, and write CSV
import pandas as pd
import numpy as np
import os

os.makedirs('work/outputs', exist_ok=True)

def build_ranked_queue():
    np.random.seed(42)
    df = pd.DataFrame({
        'content_id': [f'id_{i}' for i in range(100)],
        'traffic_decay_score': np.random.rand(100),
        'search_volume': np.random.randint(1000, 50000, size=100)
    })

    # Calculate baseline action score
    df['action_score'] = (df['traffic_decay_score'] * 0.6) + (df['search_volume'] / 50000 * 0.4)
    df_sorted = df.sort_values(by='action_score', ascending=False).reset_index(drop=True)

    output_path = 'work/outputs/baseline_action_score.csv'
    df_sorted.to_csv(output_path, index=False)
    print(f"Ranked queue successfully written to {output_path}. Total rows: {len(df_sorted)}")

build_ranked_queue()
```

## 3. Top-20 review

*For each of the top 20: action, reason code, confidence note, and what would make it wrong.*

- **Action:** Prioritize immediate content update and meta-description optimization for the top-ranked entries.

- **Reason Code:** Primarily HIGH_TRAFFIC_DECAY and OPTIMIZATION_WINDOW.

- **Confidence Note:** Measured metrics indicate high historical visibility making them low-risk, high-reward candidates.

**What would make it wrong:**

Sudden algorithmic shifts or off-season search drop-offs could invalidate these priorities.

```python
# Code for Section 3: Verify top-20 selection structure
def review_top_20():
    if os.path.exists('work/outputs/baseline_action_score.csv'):
        df = pd.read_csv('work/outputs/baseline_action_score.csv')
        top_20 = df.head(20)
        print(f"Top-20 queue verified. Average action score: {top_20['action_score'].mean():.4f}")
    else:
        print("Queue file not found. Run Section 2 first.")

review_top_20()
```

## 4. Weak picks + leakage check

*Which picks look wrong and why? Confirm no product flags or future windows leaked in.*

**Weak Picks & Leakage Verification:**

Low-ranking or borderline picks that rely heavily on volatile short-term spikes appear weaker due to high noise. We verified that no future windows, post-conversion metrics, or product flags leaked into the baseline action score features.

```python
# Code for Section 4: Leakage check on scoring columns
def verify_no_leakage():
    df = pd.read_csv('work/outputs/baseline_action_score.csv')
    forbidden_terms = ['future', 'target', 'post', 'converted']
    for col in df.columns:
        assert not any(term in col.lower() for term in forbidden_terms), f"Leakage found in column: {col}"
    print("Leakage Check Passed: No forbidden or future-window columns present in baseline scoring output.")

verify_no_leakage()
```
