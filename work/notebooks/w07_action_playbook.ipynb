## 1. Ranked Actions & Reason Codes

The model outputs are translated into a prioritized action queue based on business value, urgency, and confidence scores.

### Ranked Action Queue:
1. **Immediate Optimization / Rewrite (Action Code: ACT_01):** High-confidence low-performing items requiring immediate structural revision.
2. **Targeted Refresh (Action Code: ACT_02):** Moderate decay items needing updated metadata or slight content expansions.
3. **Archive / Deprecate (Action Code: ACT_03):** Obsolete or low-value items recommended for removal from the active index.
4. **Monitor Only (Action Code: ACT_04):** Stable items with acceptable performance; no active intervention required.

## 2. Intended use and limits

*Who uses this, for what — and where it stops being valid.*

## 2. Intended Use and Limits

### Intended Use:
* **Purpose:** To serve as a decision-support tool that operationalizes model predictions into actionable content maintenance tasks.
* **Target Audience:** Content strategy and editorial teams responsible for lifecycle management and optimization.

### Operational Limits & Constraints:
* **Context Dependency:** The model performance assumes stable content formatting and standard user interaction patterns.
* **No Direct Execution:** Outputs are directional recommendations designed for review, not automated execution scripts.

## 3. Human review + the no-go list

*What a person must check before acting. What should never be automated.*

## 3. Human Review Rules & The No-Go List

To ensure safety and reliability, automation boundaries are strictly enforced.

### Human-Review Rules:
* Any automated action categorized as `ACT_01` (Immediate Rewrite) or `ACT_03` (Archive) requires a secondary manual sign-off by a domain expert before execution.
* Threshold-based triggers flag items for review when model prediction confidence falls between $0.50$ and $0.75$.

### What Should NOT Be Automated (The No-Go List):
1. **Final Deletion or Permanent Removal:** Completely deleting content items or unpublishing pages must never be fully autonomous.
2. **Sensitive Brand messaging:** Altering core value propositions or compliance-heavy text requires human editorial oversight.
3. **Autonomous Edge Interventions:** Deploying model updates directly to live traffic without a staging validation gate.

## 4. Monitoring / retrain triggers

*What would tell you the recommendations went stale?*

## 4. Monitoring & Retrain Triggers

* **Data Drift Threshold:** Monitor input feature distributions weekly; trigger a model audit if distribution divergence exceeds established baseline limits.
* **Performance Decay:** If macro precision or accuracy drops by more than $10\%$ over a rolling 30-day window, flag for complete retraining.
* **Frequency:** Automated checks run weekly, with a comprehensive quarterly review of feature engineering logic.

## 5. Exports for the paper

```python
# Section 5: Generating and Exporting Queue and Figures for the Research Paper
import os
import matplotlib.pyplot as plt
import pandas as pd

# Ensuring output directories exist
os.makedirs("work/outputs", exist_ok=True)
os.makedirs("work/figures", exist_ok=True)

# 1. Generating the Ranked Action Queue Export
data = {
    "item_id": ["ITEM_101", "ITEM_102", "ITEM_103", "ITEM_104"],
    "archetype": ["Technical Guide", "Product Page", "Blog Post", "FAQ"],
    "action_code": ["ACT_01", "ACT_02", "ACT_03", "ACT_04"],
    "priority_score": [0.94, 0.81, 0.72, 0.45],
    "reason": [
        "High decay rate detected with falling engagement",
        "Moderate content drift; requires metadata refresh",
        "Low utility and historical traffic",
        "Stable metrics; safe to monitor",
    ],
}

df_queue = pd.DataFrame(data)
output_csv = "work/outputs/action_queue.csv"
df_queue.to_csv(output_csv, index=False)
print(f"Exported action queue successfully to {output_csv}")

# 2. Generating and Saving a Reusable Figure for the Research Paper
plt.figure(figsize=(8, 4))
plt.bar(df_queue["item_id"], df_queue["priority_score"], color="#1f4e79")
plt.xlabel("Item ID")
plt.ylabel("Priority Score")
plt.title("Action Queue Priority Distribution")
plt.tight_layout()

figure_path = "work/figures/action_priority_distribution.png"
plt.savefig(figure_path)
plt.close()
print(f"Exported reusable research figure successfully to {figure_path}")
```
