## 1. Method choice and why

**Method Selection:**
We selected a Gradient Boosting Classifier (LightGBM) as our primary modeling approach. This method fits our data engineering lane because it efficiently handles heavy-tailed numerical distributions, naturally manages mixed data types, and captures non-linear feature interactions without extensive manual scaling or tuning.

```python
# Code for Section 1: Method selection and initialization check
import lightgbm as lgb
import numpy as np
import pandas as pd

def initialize_model():
    model = lgb.LGBMClassifier(
        n_estimators=100,
        learning_rate=0.05,
        random_state=42,
        verbosity=-1
    )
    print("LightGBM baseline model successfully initialized for modeling lane.")
    return model

model = initialize_model()
```

## 2. Split design

*Grouped by client? Time-aware? Say why this split is honest for your question.*

**Time-Aware Split Design:**
To prevent data leakage and ensure an honest evaluation, we employed a time-aware chronological split rather than random sampling. Training data strictly precedes the evaluation window, mirroring real-world deployment where future outcomes cannot be seen during inference.

```python
# Code for Section 2: Time-aware split simulation
def simulate_time_split(df, time_col='timestamp', split_threshold='2026-06-01'):
    df[time_col] = pd.to_datetime(df[time_col])
    train_set = df[df[time_col] < split_threshold]
    val_set = df[df[time_col] >= split_threshold]
    print(f"Train set size: {len(train_set)}, Validation set size: {len(val_set)}")
    return train_set, val_set

mock_df = pd.DataFrame({
    'timestamp': pd.date_range(start='2026-05-01', periods=100, freq='D'),
    'feature_val': np.random.rand(100),
    'target': np.random.choice([0, 1], size=100)
})
simulate_time_split(mock_df)
```
